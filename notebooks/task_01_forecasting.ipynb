```python
import pandas as pd

# Load the sessions sheet
df = pd.read_excel('../data/ev_charging_dataset.xlsx', sheet_name='sessions')

# Convert timestamp and extract date
df['start_timestamp'] = pd.to_datetime(df['start_timestamp'])
df['date'] = df['start_timestamp'].dt.date

# Aggregate daily demand (kWh)
daily_demand = df.groupby('date')['energy_kwh'].sum().reset_index()
daily_demand.columns = ['ds', 'y'] # Rename for Prophet compatibility
daily_demand['ds'] = pd.to_datetime(daily_demand['ds'])

print(daily_demand.head())
```
